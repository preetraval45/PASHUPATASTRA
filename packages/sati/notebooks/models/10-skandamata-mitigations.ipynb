# Skandamata — from observed behaviour to the mitigations that defend against it

Model 5 of the nine: the protective response. **Data:** MITRE ATT&CK procedure examples, and ATT&CK's own `mitigates` links as ground truth. **Method:** a from-scratch multi-label network over hashed n-grams. **Split:** by source, so test behaviours describe attackers never seen in training. **Baselines:** popularity, BM25 direct and BM25 two-hop via techniques.

It recommends; Dharma authorises (rule 2). Results: `docs/research/models/skandamata.md`.
