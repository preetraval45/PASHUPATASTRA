# Mahagauri — has the server really returned to normal?

Model 8 of the nine: Verify. **Data:** SMD's real incidents. For each one, from onset until an hour after it ends, a minute-by-minute "recovered yet?"; all clear after three consecutive yeses.

**The error that matters** is a *premature all-clear*: declaring an incident over while it continues. Thresholds are set on the training groups to keep it at or below 5%. **Baselines:** back within 3σ of the pre-incident hour, and a fixed timer.

Results: `docs/research/models/mahagauri.md`.
