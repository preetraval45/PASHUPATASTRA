# Kalaratri on BGL — supercomputer logs, split by time

Model 7 of the nine, the harder dataset. **Data:** LogHub BGL, 4.7M BlueGene/L lines (research use only). **Method:** templates by masking every token with a digit, windows of 100 lines, template counts into LightGBM. **Split:** by time, never shuffled. **Baselines:** PCA (Xu et al. 2009) and the unseen-template rule.

Results: `docs/research/models/kalaratri-bgl.md`.
